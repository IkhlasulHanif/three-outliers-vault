# Zero-shot tweet topic labelling with GPT-4o through LangChain

```python
DATA_DIR = "../data"
```

```python
import os
from langchain.memory import ConversationBufferMemory
from langchain.chains import ConversationChain
from langchain.prompts import ChatPromptTemplate
from langchain_openai import OpenAI, ChatOpenAI

import pandas as pd
from sklearn.model_selection import train_test_split
from tqdm import tqdm
```

```python
chat = ChatOpenAI(
    openai_api_key=os.environ["OPENAI_API_KEY"],
    temperature=0,
    streaming=True,
    # callbacks=[StreamingStdOutCallbackHandler()],
    model_name="gpt-4o",
)
```

```python
chat.model_name
```

## Example

```python
chat_template = ChatPromptTemplate.from_template("""
Diberikan sebuah tweet, anda harus mengklasifikasikan apakah tweet ini termasuk kategori apa? Pilih salah satu kategori diantara 8 kategori berikut:
1. Politik
2. Sosial Budaya
3. Ekonomi
4. Pertahanan dan Keamanan
5. Ideologi
6. Sumber Daya Alam
7. Demografi
8. Geografi

tweet:
{text}

label?

cukup jawab dengan nama kategorinya tidak perlu menjawab apa-apa lagi.
contoh jawaban: "Kategori"
""")
```

```python
memory = ConversationBufferMemory()
conversation = ConversationChain(
    llm=chat,
    memory = memory,
    verbose=True
)
```

```python
ex1 = chat_template.format_messages(text="Kunjungan Prabowo ini untuk meresmikan dan menyerahkan proyek bantuan air bersih di lima titik. #IndonesiaSentris #IndonesiaHijau #02Melanjutkan #AnakMudaIndonesiaEmas Prabowo Subianto")
```

```python
conversation.predict(input=ex1)
```

## Real Deal

```python
train_df = pd.read_csv(f'{DATA_DIR}/train.csv', sep=";")
train_df.head()
```

```python
train_df.shape
```

```python
def drop_duplicated(df):
    print("Df size:", train_df.shape)
    train_df.drop_duplicates(subset=['text', 'label'], keep='first', inplace=True)
    print("After dropping literal duplicated rows, df size: ", train_df.shape)
    train_df.drop_duplicates(subset=['text'], keep=False, inplace=True)
    print("After dropping text duplicated rows, df size: ", train_df.shape)
    
drop_duplicated(train_df)
```

```python
train_df, val_df = train_test_split(train_df, train_size=0.2, stratify=train_df.label, random_state=42)
train_df.shape
```

```python
train_df.label.value_counts()
```

```python
# drop randomly 70% of the data that has label = Politik
train_df = train_df.drop(train_df[train_df.label == "Politik"].sample(frac=0.9, random_state=42).index)
train_df.label.value_counts()
```

```python
train_df.shape
```

```python
memory = ConversationBufferMemory()
conversation = ConversationChain(
    llm=chat,
    memory = memory,
    verbose=False
)

for i, (idx, row) in tqdm(enumerate(train_df.iterrows())):
    if i == 0:
        inp = chat_template.format_messages(text=row.text)
        res = conversation.predict(input=inp)
    else:
        res = conversation.predict(input=row.text)
    train_df.loc[idx, 'predicted_label'] = str(res)
    train_df.to_csv(f'{DATA_DIR}/train_lc.csv', index=False)
train_df.head()
```

```python
train_lc = pd.read_csv(f'{DATA_DIR}/train_lc.csv')
train_lc.dropna(subset=['predicted_label'], inplace=True)
train_lc.shape
```

```python
train_lc.head()
```

```python
from sklearn.metrics import classification_report

print(classification_report(train_lc.label, train_lc.predicted_label))
```
