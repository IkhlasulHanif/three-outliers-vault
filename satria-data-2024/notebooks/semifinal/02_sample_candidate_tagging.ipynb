# Splitting

````python
def mapping_anies(content):
  anies_keys= ['anies', 'abah']
  for anies_key in anies_keys:
    if anies_key in content.lower():
      return 1
  return 0

def mapping_prabowo(content):
  prabowo_keys= ['prabowo', 'bowo', 'gbran', 'samsul']
  for prabowo_key in prabowo_keys:
    if prabowo_key in content.lower():
      return 1
  return 0

def mapping_ganjar(content):
  ganjar_keys= ['ganjar']
  for ganjar_key in ganjar_keys:
    if ganjar_key in content.lower():
      return 1
  return 0
````

````python
df['is_anies'] = df['content'].apply(mapping_anies)
df['is_prabowo'] = df['content'].apply(mapping_prabowo)
df['is_ganjar'] = df['content'].apply(mapping_ganjar)
````
