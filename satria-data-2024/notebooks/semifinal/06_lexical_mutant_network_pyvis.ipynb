```python
import pandas as pd
import networkx as nx
from pyvis.network import Network
import pickle
```

```python
graph_df = pd.read_csv(f"{DATA_DIR}/graph_df.csv")
graph_df.rename(columns={"sim": "value"}, inplace=True)
df = pd.read_csv(f"{DATA_DIR}/df_lite.csv")
content = df["content_cleaned"].to_dict()
with open(f"{DATA_DIR}/communities.pkl", "rb") as f:
    communities = pickle.load(f)
```

```python
# filter
graph_df = graph_df.loc[graph_df["source"].isin([1, 2])]
content = {k: v for k, v in content.items() if k in [1, 2]}
communities = {k: v for k, v in communities.items() if k in [1, 2]}
print(graph_df.source.unique())
print(content.keys())
print(communities.keys())
```

```python
communities
```

```python
# Create a graph from a pandas dataframe
G = nx.from_pandas_edgelist(
    graph_df,
    source="source",
    target="target",
    edge_attr="value",
    create_using=nx.Graph()
)
```

```python
# Create a pyvis network
net = Network(
    notebook=True,
    width="100%",
    height="1000px",
    bgcolor='white',
    font_color='black',
    select_menu=True,
    filter_menu=True
)

node_degree = dict(G.degree)

# Setting up node size attribute
nx.set_node_attributes(G, node_degree, 'size')
nx.set_node_attributes(G, communities, 'color')
nx.set_node_attributes(G, content, 'title')

print("Generating network...")
net.from_nx(G, show_edge_weights=True)
net.toggle_physics(True)
net.show_buttons(filter_=['physics'])
net.show("twitter_pemilu.html")
```
