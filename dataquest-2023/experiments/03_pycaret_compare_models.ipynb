```python
import gdown
from pycaret.regression import *
import pandas as pd

DATA_DIR = "../data"
```

```python
url = 'https://drive.google.com/u/1/uc?id=1aHehuaIz-7caH83d6DWmOKsFkz6w9JrQ'
output = f"{DATA_DIR}/train_cleaned.csv"
gdown.download(url, output, quiet=False)
```

```python
train_df = pd.read_csv(f"{DATA_DIR}/train_cleaned.csv")
train_df.head()
```

```python
s = setup(
    data=train_df,
    target="rain_1h",
    session_id=42,
    use_gpu=True
)
```

```python
best_model = compare_models()
```

```python
evaluate_model(best_model)
```

```python
predict_model(best_model)
```
