```python
from google_play_scraper import app, Sort, reviews_all
import pandas as pd
import numpy as np
import json, os, uuid

play_reviews = reviews_all(
        "com.indivara.jneone",
        sleep_milliseconds=0, # defaults to 0
        lang='id', # defaults to 'en'
        country='id', # defaults to 'us'
        sort=Sort.NEWEST, # defaults to Sort.MOST_RELEVANT
    )

raw_reviews_df = pd.DataFrame(np.array(play_reviews),columns=['review'])
reviews_df = raw_reviews_df.join(pd.DataFrame(raw_reviews_df.pop('review').tolist()))

reviews_df.drop(columns={'userImage', 'reviewCreatedVersion'},inplace = True)
reviews_df.rename(columns= {'score': 'rating','userName': 'user_name', 'reviewId': 'review_id', 'content': 'review_description', 'at': 'review_date', 'replyContent': 'developer_response', 'repliedAt': 'developer_response_date', 'thumbsUpCount': 'thumbs_up'},inplace = True)
reviews_df.insert(loc=0, column='source', value='Google Play')
reviews_df.insert(loc=3, column='review_title', value=None)
reviews_df['laguage_code'] = 'id'
reviews_df['country_code'] = 'id'
```
